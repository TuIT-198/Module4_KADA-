```python
import pandas as pd
import numpy as np

df = pd.read_csv('../data/Superstore.csv', encoding='latin1')
print('Original Shape:', df.shape)
df.head()
```

```python
# 1. Data Cleaning & Feature Engineering
df['Order Date'] = pd.to_datetime(df['Order Date'], format='%d-%m-%Y')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], format='%d-%m-%Y')
df['Shipping Days'] = (df['Ship Date'] - df['Order Date']).dt.days
df['Postal Code'] = df['Postal Code'].astype(str).str.split('.').str[0].str.zfill(5)
df['Product Name'] = df['Product Name'].str.replace('�', ' ')
if 'Row ID' in df.columns:
    df = df.drop(columns=['Row ID'])
```

```python
# 2. Data Validation Checks
print('Total Missing Values:', df.isnull().sum().sum())
print('Duplicate Rows:', df.duplicated().sum())
print('Invalid Shipping Dates:', (df['Ship Date'] < df['Order Date']).sum())
print('Invalid Sales/Quantity:', (df['Sales'] <= 0).sum() + (df['Quantity'] <= 0).sum())
```

```python
# 3. Data Analyst Summary Statistics
df.info()
df[['Sales', 'Quantity', 'Discount', 'Profit', 'Shipping Days']].describe().T[['mean', 'std', 'min', '50%', 'max']]
```
